```python
from langchain_community.document_loaders import TextLoader
loader=TextLoader("mytext.txt")
text_docs=loader.load()
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
splitter=RecursiveCharacterTextSplitter(chunk_size=100,chunk_overlap=30)
final_docs=splitter.split_documents(text_docs)
```

```python
from langchain_community.embeddings import OllamaEmbeddings
embeddings=OllamaEmbeddings(model="gemma2:2b")
```

```python
from langchain_community.vectorstores import FAISS
vectordb=FAISS.from_documents(documents=final_docs,embedding=embeddings)
vectordb.save_local("faiss_index")
```
